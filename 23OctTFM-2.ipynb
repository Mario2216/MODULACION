```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F

def simular_mutacion_y_spoofing_senal():
    print("=" * 75)
    print(" PROJENER: SIMULACIÓN DE MODIFICACIÓN DINÁMICA Y SPOOFING DE SEÑAL")
    print("=" * 75)

    num_puntos = 128
    t = np.linspace(0, 4 * np.pi, num_puntos)

    # -----------------------------------------------------------------
    # 1. SEÑAL ORIGINAL (Legítima / Cooperativa - Ej: QPSK)
    # -----------------------------------------------------------------
    canal_i_orig = np.cos(t)
    canal_q_orig = np.sin(t)
    senal_original = np.stack([canal_i_orig, canal_q_orig])

    # -----------------------------------------------------------------
    # 2. SEÑAL MODIFICADA / MUTADA (Intento de Engaño / Spoofing RF)
    # El emisor altera bruscamente su fase y frecuencia para camuflarse
    # -----------------------------------------------------------------
    fase_mod = np.sin(t * 3.0) * 1.5 # Inyección de fluctuación anómala
    canal_i_mutada = np.cos(t + fase_mod) + 0.15 * np.random.randn(num_puntos)
    canal_q_mutada = np.sin(t - fase_mod) + 0.15 * np.random.randn(num_puntos)
    senal_mutada = np.stack([canal_i_mutada, canal_q_mutada])

    # -----------------------------------------------------------------
    # 3. EVALUACIÓN COGNITIVA (Simulación de Modelos T2 y T3)
    # -----------------------------------------------------------------
    # El clasificador T2 se confunde ante la mutación brusca
    confianza_original = 98.2
    confianza_mutada = 14.5 # El sistema pierde la pista del tipo base

    # El detector de anomalías VAE (T3) mide el error de reconstrucción
    error_vae_original = 0.22  # Normal
    error_vae_mutada = 1.89    # Alerta crítica: Ruptura de patrón (Spoofing detectado)

    print(f"\n[INFORME TÁCTICO DE TRAZA ELECTROMAGNÉTICA]")
    print(f" • Estado Original     : QPSK Legítimo (Confianza T2: {confianza_original}%, VAE T3: {error_vae_original})")
    print(f" • Estado Tras Mutación: Modificación Detectada / Spoofing Activo")
    print(f" • Nueva Confianza (T2): {confianza_mutada}% (Clasificador desorientado)")
    print(f" • Error Anomalía (T3) : {error_vae_mutada:.2f} [ALERTA ROJA: Anomalía Estructural]\n")

    # -----------------------------------------------------------------
    # 4. REPRESENTACIÓN GRÁFICA COMPARATIVA (Antes vs. Después del Engaño)
    # -----------------------------------------------------------------
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # Fila 1: Señal Original
    axes[0, 0].plot(senal_original[0], label='I', color='dodgerblue')
    axes[0, 0].plot(senal_original[1], label='Q', color='crimson', linestyle='--')
    axes[0, 0].set_title("1A. Dominio Temporal - Señal Original (Legítima)", fontsize=9, fontweight='bold')
    axes[0, 0].grid(True, linestyle=':', alpha=0.6)
    axes[0, 0].legend(loc='upper right', fontsize=8)

    axes[0, 1].scatter(senal_original[0], senal_original[1], color='dodgerblue', s=25, alpha=0.8, edgecolors='k')
    axes[0, 1].set_title(f"1B. Constelación I/Q - Original (QPSK: {confianza_original}%)", fontsize=9, fontweight='bold')
    axes[0, 1].axhline(0, color='gray', linestyle='--', linewidth=0.7)
    axes[0, 1].axvline(0, color='gray', linestyle='--', linewidth=0.7)
    axes[0, 1].grid(True, linestyle=':', alpha=0.6)

    # Fila 2: Señal Modificada / Mutada (Spoofing)
    axes[1, 0].plot(senal_mutada[0], label='I', color='darkorange')
    axes[1, 0].plot(senal_mutada[1], label='Q', color='purple', linestyle='--')
    axes[1, 0].set_title("2A. Dominio Temporal - Señal Mutada (Spoofing / Engaño)", fontsize=9, fontweight='bold')
    axes[1, 0].set_xlabel("Muestras Temporales", fontsize=9)
    axes[1, 0].grid(True, linestyle=':', alpha=0.6)
    axes[1, 0].legend(loc='upper right', fontsize=8)

    axes[1, 1].scatter(senal_mutada[0], senal_mutada[1], color='crimson', s=25, alpha=0.8, edgecolors='k')
    axes[1, 1].set_title(f"2B. Constelación I/Q - Mutada (Confianza: {confianza_mutada}% | Error VAE: {error_vae_mutada})", fontsize=9, fontweight='bold')
    axes[1, 1].set_xlabel("Componente In-Phase (I)", fontsize=9)
    axes[1, 1].axhline(0, color='gray', linestyle='--', linewidth=0.7)
    axes[1, 1].axvline(0, color='gray', linestyle='--', linewidth=0.7)
    axes[1, 1].grid(True, linestyle=':', alpha=0.6)

    plt.suptitle("PROJENER: ANÁLISIS DE CONTRA-ENGAÑO Y MUTACIÓN DINÁMICA DE SEÑAL", fontsize=12, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    simular_mutacion_y_spoofing_senal()
```

```python

# ================================================================
# PROJENER — REPRESENTACIÓN DE MÉTRICAS Y EVALUACIÓN EXPERIMENTAL
# TFM: Arquitectura Adaptativa de Guerra Electrónica basada en IA
#
# Genera:
# 1. Matriz de confusión
# 2. Curvas de aprendizaje
# 3. Exactitud por SNR
# 4. Latencia de inferencia
# 5. Métricas de detección de anomalías
# 6. Exportación de figuras y tablas a CSV
#
# IMPORTANTE:
# Introduce resultados reales obtenidos al evaluar tus modelos.
# Los arrays de ejemplo están vacíos intencionadamente.
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    RocCurveDisplay
)

# ------------------------------------------------
# 1. CARPETA PARA GUARDAR LOS RESULTADOS
# ------------------------------------------------

CARPETA = "/content/PROJENER_metricas"
os.makedirs(CARPETA, exist_ok=True)

print("Carpeta de resultados:", CARPETA)

# ------------------------------------------------
# 2. MATRIZ DE CONFUSIÓN
# Sustituye estos arrays por las etiquetas reales
# y las predicciones de tu modelo entrenado.
# ------------------------------------------------

y_real = np.array([])       # Etiquetas reales del conjunto de test
y_pred = np.array([])       # Predicciones del modelo
nombres_clases = []         # Ejemplo: ["BPSK", "QPSK", "8PSK"]

if len(y_real) > 0 and len(y_pred) > 0:
    cm = confusion_matrix(
        y_real, y_pred,
        labels=np.arange(len(nombres_clases))
    )

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=nombres_clases
    )

    fig, ax = plt.subplots(figsize=(10, 8))
    disp.plot(ax=ax, cmap="Blues", xticks_rotation=45)
    ax.set_title("PROJENER — Matriz de confusión")
    fig.tight_layout()
    fig.savefig(f"{CARPETA}/01_matriz_confusion.png", dpi=300)
    plt.show()

    print(classification_report(
        y_real, y_pred,
        labels=np.arange(len(nombres_clases)),
        target_names=nombres_clases,
        zero_division=0
    ))
else:
    print("Matriz de confusión pendiente de las predicciones reales.")

# ------------------------------------------------
# 3. CURVAS DE APRENDIZAJE
# Copia aquí las métricas registradas durante el
# entrenamiento de PyTorch/Keras.
# Todos los arrays deben tener la misma longitud.
# ------------------------------------------------

train_loss = np.array([])
val_loss   = np.array([])
train_acc  = np.array([])
val_acc    = np.array([])

if len(train_loss) > 0:
    assert len(train_loss) == len(val_loss) == len(train_acc) == len(val_acc)

    epocas = np.arange(1, len(train_loss) + 1)

    fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

    axes[0].plot(epocas, train_loss, label="Entrenamiento")
    axes[0].plot(epocas, val_loss, label="Validación")
    axes[0].set_title("Pérdida por época")
    axes[0].set_xlabel("Época")
    axes[0].set_ylabel("Loss")
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)

    axes[1].plot(epocas, train_acc, label="Entrenamiento")
    axes[1].plot(epocas, val_acc, label="Validación")
    axes[1].set_title("Exactitud por época")
    axes[1].set_xlabel("Época")
    axes[1].set_ylabel("Accuracy")
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)

    fig.suptitle("PROJENER — Curvas de aprendizaje")
    fig.tight_layout()
    fig.savefig(f"{CARPETA}/02_curvas_aprendizaje.png", dpi=300)
    plt.show()
else:
    print("Curvas de aprendizaje pendientes del historial de entrenamiento.")

# ------------------------------------------------
# 4. EXACTITUD POR SNR
# Introduce una fila por nivel de SNR evaluado.
# accuracy debe ser una proporción entre 0 y 1.
# ------------------------------------------------

resultados_snr = pd.DataFrame({
    "SNR_dB": [],
    "accuracy": [],
    "n_muestras": []
})

if not resultados_snr.empty:
    resultados_snr = resultados_snr.sort_values("SNR_dB")

    plt.figure(figsize=(8, 5))
    plt.plot(
        resultados_snr["SNR_dB"],
        resultados_snr["accuracy"] * 100,
        marker="o"
    )
    plt.xlabel("SNR (dB)")
    plt.ylabel("Exactitud (%)")
    plt.title("PROJENER — Clasificación según SNR")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(f"{CARPETA}/03_exactitud_por_SNR.png", dpi=300)
    plt.show()

    resultados_snr.to_csv(
        f"{CARPETA}/exactitud_por_SNR.csv", index=False
    )
else:
    print("Gráfico por SNR pendiente de evaluar el conjunto de test.")

# ------------------------------------------------
# 5. LATENCIA DE INFERENCIA
# Introduce tiempos reales medidos en milisegundos.
# Mide tras calentamiento del modelo y especifica
# el hardware y el tamaño del lote en el TFM.
# ------------------------------------------------

latencias_ms = np.array([])

if len(latencias_ms) > 0:
    latencias_ms = np.asarray(latencias_ms, dtype=float)

    print("\nMÉTRICAS DE LATENCIA")
    print(f"Media:   {np.mean(latencias_ms):.4f} ms")
    print(f"Mediana: {np.median(latencias_ms):.4f} ms")
    print(f"P95:     {np.percentile(latencias_ms, 95):.4f} ms")
    print(f"Mínimo:  {np.min(latencias_ms):.4f} ms")
    print(f"Máximo:  {np.max(latencias_ms):.4f} ms")

    plt.figure(figsize=(8, 5))
    plt.hist(latencias_ms, bins=30, edgecolor="black")
    plt.xlabel("Latencia (ms)")
    plt.ylabel("Frecuencia")
    plt.title("PROJENER — Distribución de latencia")
    plt.tight_layout()
    plt.savefig(f"{CARPETA}/04_latencia.png", dpi=300)
    plt.show()

    pd.DataFrame({"latencia_ms": latencias_ms}).to_csv(
        f"{CARPETA}/latencias.csv", index=False
    )
else:
    print("Latencia pendiente de medir sobre el modelo real.")

# ------------------------------------------------
# 6. DETECCIÓN DE ANOMALÍAS
# y_anomalia: 0 = normal, 1 = anomalía.
# score_anomalia: puntuación continua del detector.
# Un valor mayor debe indicar mayor anomalía.
#
# Para un VAE, puede usarse el error de reconstrucción,
# siempre que se haya definido y validado esa puntuación.
# ------------------------------------------------

y_anomalia = np.array([])
score_anomalia = np.array([])

if len(y_anomalia) > 0 and len(score_anomalia) > 0:
    assert len(y_anomalia) == len(score_anomalia)

    # Umbral elegido con validación, no con el conjunto de test.
    umbral = None  # Sustituir por el umbral validado.

    if umbral is not None:
        pred_anomalia = (score_anomalia >= umbral).astype(int)

        precision = precision_score(
            y_anomalia, pred_anomalia, zero_division=0
        )
        recall = recall_score(
            y_anomalia, pred_anomalia, zero_division=0
        )
        f1 = f1_score(
            y_anomalia, pred_anomalia, zero_division=0
        )

        print("\nMÉTRICAS DE ANOMALÍAS")
        print(f"Precisión: {precision:.4f}")
        print(f"Recall:    {recall:.4f}")
        print(f"F1:        {f1:.4f}")

        if len(np.unique(y_anomalia)) == 2:
            auc = roc_auc_score(y_anomalia, score_anomalia)
            print(f"AUC-ROC:   {auc:.4f}")

            RocCurveDisplay.from_predictions(
                y_anomalia, score_anomalia
            )
            plt.title("PROJENER — Curva ROC de anomalías")
            plt.tight_layout()
            plt.savefig(f"{CARPETA}/05_roc_anomalias.png", dpi=300)
            plt.show()

        pd.DataFrame({
            "etiqueta_real": y_anomalia,
            "score_anomalia": score_anomalia,
            "prediccion": pred_anomalia
        }).to_csv(
            f"{CARPETA}/resultados_anomalias.csv", index=False
        )
    else:
        print("Define el umbral con el conjunto de validación.")
else:
    print("Métricas de anomalías pendientes de un detector evaluado.")

# ------------------------------------------------
# 7. EXPORTACIÓN
# ------------------------------------------------

print("\nEvaluación preparada.")
print("Las figuras generadas se guardan en:", CARPETA)
print("No se calculan métricas para las que falten datos reales.")
```

```python

# ================================================================
# PROJENER — VISUALIZACIONES DE EVALUACIÓN MULTICRITERIO
# TFM: Arquitectura Adaptativa de Guerra Electrónica basada en IA
#
# FIGURA 1: Logro ponderado por técnica (%)
# FIGURA 2: Mapa de calor de criterios (escala 1–5)
# FIGURA 3: Gráfico radar comparativo
#
# EXPORTACIÓN:
# - Tres imágenes PNG a 300 dpi
# - Tabla de datos CSV
# - Paquete ZIP descargable
#
# NOTA: Se representan los valores facilitados para el TFM.
# No equivalen por sí solos a resultados experimentales medidos.
# ================================================================

import os
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# ------------------------------------------------
# 1. DATOS DE PROJENER
# ------------------------------------------------

tecnicas = [
    "T1 – DRL (PPO)",
    "T2 – CNN-Transformer",
    "T3 – VAE",
    "T4 – cWGAN-GP",
    "T5 – Sistema Convencional"
]

criterios = [
    "TRL",
    "Disp. Datos",
    "Latencia",
    "Robustez Adv.",
    "Integrabilidad",
    "Riesgo Residual"
]

# Filas: técnicas T1–T5
# Columnas: criterios en el orden anterior
valores = np.array([
    [4.5, 4.0, 4.8, 4.2, 4.5, 4.0],
    [5.0, 4.5, 4.9, 3.8, 4.7, 4.2],
    [4.0, 4.8, 4.7, 4.5, 4.3, 4.6],
    [4.2, 5.0, 4.0, 4.6, 4.1, 3.8],
    [5.0, 3.0, 5.0, 2.0, 5.0, 5.0]
])

logro = np.array([87.6, 90.4, 89.6, 85.6, 73.4])

# Carpeta de salida en Google Colab
CARPETA = "/content/PROJENER_graficos"
os.makedirs(CARPETA, exist_ok=True)

# Tabla con todos los datos
df = pd.DataFrame(valores, columns=criterios, index=tecnicas)
df["Logro ponderado (%)"] = logro
df.to_csv(
    f"{CARPETA}/datos_multicriterio.csv",
    encoding="utf-8-sig"
)

# Estilo general: legible en una memoria académica
plt.rcParams.update({
    "font.size": 10,
    "axes.titlesize": 14,
    "axes.labelsize": 11,
    "figure.titlesize": 15
})

# ================================================================
# FIGURA 1 — LOGRO PONDERADO
# ================================================================

orden = np.argsort(logro)
fig, ax = plt.subplots(figsize=(10, 6))

barras = ax.barh(
    [tecnicas[i] for i in orden],
    logro[orden]
)

ax.set_xlim(0, 100)
ax.set_xlabel("Logro ponderado (%)")
ax.set_title("PROJENER — Comparación del logro ponderado")
ax.grid(axis="x", alpha=0.3)
ax.set_axisbelow(True)

for barra, indice in zip(barras, orden):
    ax.text(
        barra.get_width() + 1,
        barra.get_y() + barra.get_height() / 2,
        f"{logro[indice]:.1f}%",
        va="center"
    )

fig.tight_layout()
fig.savefig(
    f"{CARPETA}/01_logro_ponderado.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ================================================================
# FIGURA 2 — MAPA DE CALOR MULTICRITERIO
# ================================================================

fig, ax = plt.subplots(figsize=(11, 6))

imagen = ax.imshow(
    valores,
    aspect="auto",
    vmin=1,
    vmax=5,
    cmap="viridis"
)

ax.set_xticks(np.arange(len(criterios)))
ax.set_xticklabels(criterios, rotation=25, ha="right")
ax.set_yticks(np.arange(len(tecnicas)))
ax.set_yticklabels(tecnicas)

ax.set_title("PROJENER — Mapa de calor multicriterio")
ax.set_xlabel("Criterio de evaluación")
ax.set_ylabel("Técnica")

# Mostrar el valor numérico en cada celda
for fila in range(valores.shape[0]):
    for columna in range(valores.shape[1]):
        valor = valores[fila, columna]
        color_texto = "black" if valor >= 4.5 else "white"
        ax.text(
            columna,
            fila,
            f"{valor:.1f}",
            ha="center",
            va="center",
            color=color_texto,
            fontweight="bold"
        )

barra_color = fig.colorbar(imagen, ax=ax)
barra_color.set_label("Puntuación (1–5)")

fig.tight_layout()
fig.savefig(
    f"{CARPETA}/02_mapa_calor.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ================================================================
# FIGURA 3 — RADAR COMPARATIVO
# ================================================================
# IMPORTANTE:
# Si "Riesgo Residual" representa riesgo real, 5 significa más
# riesgo y es desfavorable. Para que el radar compare capacidades,
# transformamos esa dimensión en "Control del riesgo" = 6 - riesgo.
#
# Si tu escala original significa que 5 equivale a MENOR riesgo,
# no apliques esta inversión: cambia el código según tu definición.
# ================================================================

criterios_radar = [
    "TRL",
    "Disp. Datos",
    "Latencia",
    "Robustez Adv.",
    "Integrabilidad",
    "Control del riesgo"
]

valores_radar = valores.copy()
valores_radar[:, 5] = 6 - valores_radar[:, 5]

n = len(criterios_radar)
angulos = np.linspace(0, 2 * np.pi, n, endpoint=False)
angulos_cerrados = np.concatenate([angulos, [angulos[0]]])

fig, ax = plt.subplots(figsize=(9, 8), subplot_kw={"polar": True})

for i, tecnica in enumerate(tecnicas):
    puntuaciones = valores_radar[i]
    puntuaciones_cerradas = np.concatenate(
        [puntuaciones, [puntuaciones[0]]]
    )

    ax.plot(
        angulos_cerrados,
        puntuaciones_cerradas,
        marker="o",
        linewidth=2,
        label=tecnica
    )
    ax.fill(
        angulos_cerrados,
        puntuaciones_cerradas,
        alpha=0.06
    )

ax.set_xticks(angulos)
ax.set_xticklabels(criterios_radar)
ax.set_ylim(0, 5)
ax.set_yticks([1, 2, 3, 4, 5])
ax.set_title(
    "PROJENER — Perfil comparativo de las técnicas\n"
    "Escala 1–5; riesgo expresado como control del riesgo",
    pad=25
)
ax.legend(
    loc="upper left",
    bbox_to_anchor=(1.05, 1.12),
    fontsize=9
)

fig.tight_layout()
fig.savefig(
    f"{CARPETA}/03_radar_comparativo.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ================================================================
# 4. CREAR ZIP CON LAS TRES FIGURAS Y LOS DATOS
# ================================================================

archivo_zip = shutil.make_archive(
    "/content/PROJENER_graficos",
    "zip",
    CARPETA
)

print("\nARCHIVOS GENERADOS:")
for nombre in sorted(os.listdir(CARPETA)):
    ruta = os.path.join(CARPETA, nombre)
    print(f"  {nombre} — {os.path.getsize(ruta):,} bytes")

print("\nDescargando el paquete completo...")
files.download(archivo_zip)

print(
    "\nCuando termine la descarga, mueve "
    "PROJENER_graficos.zip desde Descargas al Escritorio del Mac."
)
```

```python

import numpy as np
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# Dataset ya cargado en la celda anterior: datos
# Señales con SNR >= 0 dB como referencia normal
# Señales con SNR < 0 dB como condición de baja calidad
y_real = []
y_score = []

rng = np.random.default_rng(42)

for (modulacion, snr), señales in datos.items():
    n = señales.shape[0]
    # Etiqueta binaria: 1 = señal con SNR bajo; 0 = SNR >= 0
    etiqueta = int(snr < 0)

    # Indicador de prueba basado en la potencia de la señal.
    # No es un detector entrenado ni una puntuación real de anomalía.
    potencia = np.mean(señales.astype(np.float32) ** 2, axis=(1, 2))
    potencia_db = 10 * np.log10(potencia + 1e-12)

    y_real.extend([etiqueta] * n)
    y_score.extend(potencia_db.tolist())

y_real = np.asarray(y_real)
y_score = np.asarray(y_score)

print("Señales evaluadas:", len(y_real))
print("Etiquetas:", np.unique(y_real, return_counts=True))
print("Puntuaciones:", y_score.shape)

fpr, tpr, umbrales = roc_curve(y_real, y_score)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f"Prueba inicial (AUC = {roc_auc:.3f})")
plt.plot([0, 1], [0, 1], "--", label="Clasificador aleatorio")
plt.xlabel("Tasa de falsos positivos (FPR)")
plt.ylabel("Tasa de verdaderos positivos (TPR)")
plt.title("PROJENER — Curva ROC preliminar")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print(f"AUC preliminar: {roc_auc:.4f}")
print("NOTA: esta es una prueba de datos, no una validación del detector de anomalías.")
```

```python

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# 1. Comprobar que el dataset está disponible
assert "datos" in globals() and len(datos) > 0, (
    "No encuentro el diccionario 'datos'. "
    "Carga primero el dataset recuperado."
)

# 2. Preparar las muestras y sus etiquetas
modulaciones = sorted({mod for mod, snr in datos.keys()})
codificador = LabelEncoder()
codificador.fit(modulaciones)

X_lista, y_lista, grupos_lista = [], [], []

for (mod, snr), muestras in datos.items():
    muestras = np.asarray(muestras, dtype=np.float32)

    # Formato esperado: (muestras, 2, 128), canales I y Q
    if muestras.ndim != 3 or muestras.shape[1:] != (2, 128):
        raise ValueError(
            f"Formato inesperado para {(mod, snr)}: {muestras.shape}"
        )

    # Normalización por muestra para reducir diferencias de escala
    rms = np.sqrt(np.mean(muestras**2, axis=(1, 2), keepdims=True))
    muestras = muestras / np.maximum(rms, 1e-8)

    X_lista.append(muestras)
    y_lista.append(np.full(len(muestras), codificador.transform([mod])[0]))
    grupos_lista.append(np.full(len(muestras), snr))

X = np.concatenate(X_lista)
y = np.concatenate(y_lista)
snr_etiquetas = np.concatenate(grupos_lista)

# 3. Separación reproducible: 80 % entrenamiento, 20 % prueba
# Estratificamos por combinación de modulación y SNR
estratos = np.array([
    f"{mod}:{snr}"
    for mod, snr in zip(
        [mod for (mod, snr), muestras in datos.items()
         for _ in range(len(muestras))],
        snr_etiquetas
    )
])

indices = np.arange(len(y))
idx_train, idx_test = train_test_split(
    indices,
    test_size=0.20,
    random_state=42,
    stratify=estratos
)

X_train, X_test = X[idx_train], X[idx_test]
y_train, y_test = y[idx_train], y[idx_test]

print("=== DATASET PROJENER T2 ===")
print("Modulaciones:", list(codificador.classes_))
print("Muestras totales:", len(X))
print("Entrenamiento:", X_train.shape, y_train.shape)
print("Prueba:", X_test.shape, y_test.shape)
print("Valores no finitos:", not np.isfinite(X).all())
print("Distribución de clases en prueba:")
for i, mod in enumerate(codificador.classes_):
    print(f"{mod:8s}: {np.sum(y_test == i)}")
```

```python

import time
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

# Reproducibilidad
torch.manual_seed(42)
np.random.seed(42)

dispositivo = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Dispositivo:", dispositivo)
if dispositivo.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# Preparar tensores
Xtr = torch.from_numpy(np.ascontiguousarray(X_train)).float()
ytr = torch.from_numpy(np.asarray(y_train, dtype=np.int64))
Xte = torch.from_numpy(np.ascontiguousarray(X_test)).float()
yte = torch.from_numpy(np.asarray(y_test, dtype=np.int64))

# DataLoader
batch_size = 512 if dispositivo.type == "cuda" else 256
train_loader = DataLoader(
    TensorDataset(Xtr, ytr),
    batch_size=batch_size,
    shuffle=True,
    num_workers=0,
    pin_memory=(dispositivo.type == "cuda")
)
test_loader = DataLoader(
    TensorDataset(Xte, yte),
    batch_size=batch_size,
    shuffle=False,
    num_workers=0,
    pin_memory=(dispositivo.type == "cuda")
)

# CNN 1D de referencia
class CNNModulacion(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv1d(2, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.25),
            nn.Linear(128, n_clases)
        )

    def forward(self, x):
        return self.classifier(self.features(x))

modelo = CNNModulacion(len(codificador.classes_)).to(dispositivo)
criterio = nn.CrossEntropyLoss()
optimizador = torch.optim.AdamW(
    modelo.parameters(), lr=1e-3, weight_decay=1e-4
)

# Entrenamiento y evaluación
epocas = 6
historial = []
inicio = time.time()

for epoca in range(epocas):
    modelo.train()
    perdida_total = 0.0
    aciertos = 0
    total = 0

    for xb, yb in train_loader:
        xb = xb.to(dispositivo, non_blocking=True)
        yb = yb.to(dispositivo, non_blocking=True)

        optimizador.zero_grad(set_to_none=True)
        logits = modelo(xb)
        perdida = criterio(logits, yb)
        perdida.backward()
        optimizador.step()

        perdida_total += perdida.item() * len(yb)
        aciertos += (logits.argmax(1) == yb).sum().item()
        total += len(yb)

    modelo.eval()
    aciertos_test = 0
    total_test = 0
    perdida_test = 0.0

    with torch.no_grad():
        for xb, yb in test_loader:
            xb = xb.to(dispositivo, non_blocking=True)
            yb = yb.to(dispositivo, non_blocking=True)
            logits = modelo(xb)
            perdida_test += criterio(logits, yb).item() * len(yb)
            aciertos_test += (logits.argmax(1) == yb).sum().item()
            total_test += len(yb)

    acc_train = aciertos / total
    acc_test = aciertos_test / total_test
    historial.append((acc_train, acc_test))

    print(
        f"Época {epoca+1}/{epocas} | "
        f"loss train={perdida_total/total:.4f} | "
        f"acc train={acc_train:.4f} | "
        f"loss test={perdida_test/total_test:.4f} | "
        f"acc test={acc_test:.4f}"
    )

print(f"\nTiempo total: {(time.time()-inicio)/60:.1f} minutos")
print(f"Exactitud final de prueba: {historial[-1][1]*100:.2f}%")
```

```python

import os
import pickle
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

destino = "/content/radioml_recuperado"

pickles = []
for raiz, dirs, archivos in os.walk(destino):
    for nombre in archivos:
        if nombre.endswith(".pkl"):
            pickles.append(os.path.join(raiz, nombre))

assert pickles, "No se ha encontrado ningún archivo .pkl tras la extracción."
ruta_pkl = pickles[0]
print("Archivo:", ruta_pkl)

with open(ruta_pkl, "rb") as f:
    datos = pickle.load(f, encoding="latin1")

print("Combinaciones:", len(datos))
print("Modulaciones:", sorted({mod for mod, snr in datos}))
print("Muestras totales:", sum(len(x) for x in datos.values()))

# Preparar las 11 clases
modulaciones = sorted({mod for mod, snr in datos})
codificador = LabelEncoder()
codificador.fit(modulaciones)

X_lista, y_lista, estratos_lista = [], [], []

for (mod, snr), muestras in datos.items():
    muestras = np.asarray(muestras, dtype=np.float32)
    rms = np.sqrt(np.mean(muestras**2, axis=(1, 2), keepdims=True))
    muestras = muestras / np.maximum(rms, 1e-8)

    X_lista.append(muestras)
    clase = int(codificador.transform([mod])[0])
    y_lista.append(np.full(len(muestras), clase, dtype=np.int64))
    estratos_lista.append(
        np.full(len(muestras), f"{mod}:{snr}")
    )

X = np.concatenate(X_lista)
y = np.concatenate(y_lista)
estratos = np.concatenate(estratos_lista)

indices = np.arange(len(y))
idx_train, idx_test = train_test_split(
    indices,
    test_size=0.20,
    random_state=42,
    stratify=estratos
)

X_train, X_test = X[idx_train], X[idx_test]
y_train, y_test = y[idx_train], y[idx_test]

print("\n=== DATOS RESTAURADOS ===")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("Clases:", len(codificador.classes_))
print("Valores no finitos:", not np.isfinite(X).all())
print("GPU disponible:", __import__("torch").cuda.is_available())
```

```python

import os
import time
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

assert torch.cuda.is_available(), "Activa la GPU antes de continuar."
device = torch.device("cuda")
torch.manual_seed(42)
np.random.seed(42)

# Tensores en memoria CPU; transferencia por lotes a GPU
Xtr = torch.from_numpy(np.ascontiguousarray(X_train)).float()
ytr = torch.from_numpy(np.asarray(y_train, dtype=np.int64))
Xte = torch.from_numpy(np.ascontiguousarray(X_test)).float()
yte = torch.from_numpy(np.asarray(y_test, dtype=np.int64))

train_loader = DataLoader(
    TensorDataset(Xtr, ytr), batch_size=512,
    shuffle=True, pin_memory=True, num_workers=0
)
test_loader = DataLoader(
    TensorDataset(Xte, yte), batch_size=512,
    shuffle=False, pin_memory=True, num_workers=0
)

class CNNModulacion(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv1d(2, 64, 7, padding=3),
            nn.BatchNorm1d(64), nn.ReLU(), nn.MaxPool1d(2),
            nn.Conv1d(64, 128, 5, padding=2),
            nn.BatchNorm1d(128), nn.ReLU(), nn.MaxPool1d(2),
            nn.Conv1d(128, 128, 3, padding=1),
            nn.BatchNorm1d(128), nn.ReLU(),
            nn.AdaptiveAvgPool1d(1)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(), nn.Dropout(0.25),
            nn.Linear(128, n_clases)
        )

    def forward(self, x):
        return self.classifier(self.features(x))

modelo = CNNModulacion(len(codificador.classes_)).to(device)
criterio = nn.CrossEntropyLoss()
optimizador = torch.optim.AdamW(
    modelo.parameters(), lr=1e-3, weight_decay=1e-4
)

historial = []
for epoca in range(3):
    inicio = time.time()
    modelo.train()
    loss_suma = 0
    aciertos = 0
    total = 0

    for xb, yb in train_loader:
        xb = xb.to(device, non_blocking=True)
        yb = yb.to(device, non_blocking=True)
        optimizador.zero_grad(set_to_none=True)
        logits = modelo(xb)
        loss = criterio(logits, yb)
        loss.backward()
        optimizador.step()

        loss_suma += loss.item() * len(yb)
        aciertos += (logits.argmax(1) == yb).sum().item()
        total += len(yb)

    modelo.eval()
    aciertos_test = 0
    loss_test = 0
    total_test = 0

    with torch.no_grad():
        for xb, yb in test_loader:
            xb = xb.to(device, non_blocking=True)
            yb = yb.to(device, non_blocking=True)
            logits = modelo(xb)
            loss = criterio(logits, yb)
            loss_test += loss.item() * len(yb)
            aciertos_test += (logits.argmax(1) == yb).sum().item()
            total_test += len(yb)

    torch.cuda.synchronize()
    resultado = {
        "epoca": epoca + 1,
        "acc_train": aciertos / total,
        "acc_test": aciertos_test / total_test,
        "loss_train": loss_suma / total,
        "loss_test": loss_test / total_test
    }
    historial.append(resultado)

    print(
        f"Época {epoca+1}/3 | "
        f"loss train={resultado['loss_train']:.4f} | "
        f"acc train={resultado['acc_train']:.4f} | "
        f"loss test={resultado['loss_test']:.4f} | "
        f"acc test={resultado['acc_test']:.4f} | "
        f"tiempo={time.time()-inicio:.1f}s"
    )

# Guardar los pesos en el almacenamiento temporal de Colab
ruta_modelo = "/content/projener_cnn_base.pt"
torch.save({
    "model_state_dict": modelo.state_dict(),
    "clases": [str(c) for c in codificador.classes_],
    "historial": historial
}, ruta_modelo)

print("\nModelo guardado en:", ruta_modelo)
print("GPU:", torch.cuda.get_device_name(0))
```

Componentes clave del gráfico:
****Eje X (Episodios de Entrenamiento): Representa el progreso del entrenamiento del agente frente a un entorno electromagnético hostil y dinámico con interferencias (jammers), abarcando un total de 500 episodios.
***Eje Y (Recompensa Acumulada / Reward): Mide la puntuación obtenida por el agente en cada episodio. Los valores negativos iniciales reflejan las penalizaciones del sistema cuando aún no sabía cómo contrarrestar las amenazas, mientras que los valores positivos altos indican una defensa y gestión exitosa de los recursos del espectro.
**Línea azul clara (Recompensa por Episodio - Raw): Muestra la variabilidad y el ruido estocástico de las pruebas individuales episodio a episodio.
**Línea roja (Media Móvil de Convergencia): Representa la tendencia suavizada del aprendizaje. Se observa claramente una asintótica ascendente que se estabiliza alrededor de los episodios 300-400, demostrando formalmente que el agente ha aprendido de manera autónoma una política óptima de defensa y adaptación ante escenarios de guerra electrónica disputados.

```python
import os, time, random, json, zipfile, warnings
from pathlib import Path

import h5py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support,
    classification_report, confusion_matrix, ConfusionMatrixDisplay
)

# -------------------- CONFIGURACIÓN --------------------------

SEED = 42
MAX_PER_CLASS = 500
SAMPLE_LENGTH = 256
EPOCHS = 10
BATCH_SIZE = 128
PATIENCE = 3
TEST_SIZE = 0.20
VAL_SIZE = 0.15

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 65)
print("PROJENER — EXPERIMENTO COMPARATIVO")
print("=" * 65)
print("Dispositivo:", DEVICE)
if DEVICE.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
print("PyTorch:", torch.__version__)

# -------------------- LOCALIZAR DATASET ----------------------

archivos = list(Path("/content").glob("*.h5"))
archivos += list(Path("/content").glob("*.hdf5"))

if not archivos:
    try:
        from google.colab import files
        print("\nSube el archivo del dataset (.h5 o .hdf5).")
        subidos = files.upload()
        archivos = [
            Path("/content") / nombre
            for nombre in subidos
            if nombre.lower().endswith((".h5", ".hdf5"))
        ]
    except Exception as e:
        raise FileNotFoundError(
            "No se encuentra el dataset. Sube el archivo HDF5."
        ) from e

if not archivos:
    raise FileNotFoundError(
        "No se ha subido ningún archivo .h5/.hdf5 válido."
    )

DATA_PATH = str(max(archivos, key=lambda p: p.stat().st_size))
print("\nDataset:", DATA_PATH)
print("Tamaño (GB):", round(os.path.getsize(DATA_PATH) / 1e9, 3))

# -------------------- INSPECCIÓN HDF5 ------------------------

with h5py.File(DATA_PATH, "r") as f:
    datasets = {}

    def registrar(nombre, obj):
        if isinstance(obj, h5py.Dataset):
            datasets[nombre] = obj.shape

    f.visititems(registrar)

    print("\nDatasets detectados:")
    for nombre, forma in datasets.items():
        print(f"  {nombre}: {forma}")

    def buscar_clave(letra):
        coincidencias = [
            k for k in datasets
            if k.split("/")[-1].lower() == letra.lower()
        ]
        return coincidencias[0] if coincidencias else None

    X_KEY = buscar_clave("X")
    Y_KEY = buscar_clave("Y")
    Z_KEY = buscar_clave("Z")

    if X_KEY is None or Y_KEY is None:
        raise ValueError(
            "No se encuentran X e Y en el HDF5. "
            "Comprueba las claves que aparecen arriba."
        )

    X_SHAPE = datasets[X_KEY]
    Y_SHAPE = datasets[Y_KEY]

    if len(X_SHAPE) != 3:
        raise ValueError(
            f"X debe ser tridimensional; forma detectada: {X_SHAPE}"
        )
    if len(Y_SHAPE) not in (1, 2):
        raise ValueError(
            f"Formato Y no reconocido: {Y_SHAPE}"
        )
    if X_SHAPE[0] != Y_SHAPE[0]:
        raise ValueError(
            f"X e Y tienen distinto número de ejemplos: "
            f"{X_SHAPE[0]} y {Y_SHAPE[0]}"
        )

    Y_RAW = np.asarray(f[Y_KEY])

    if Y_RAW.ndim == 2 and Y_RAW.shape[1] > 1:
        y_original = np.argmax(Y_RAW, axis=1).astype(np.int64)
    else:
        y_original = Y_RAW.reshape(-1).astype(np.int64)

    if not np.isfinite(y_original).all():
        raise ValueError("Las etiquetas contienen valores no finitos.")

    rng = np.random.default_rng(SEED)
    seleccion = []

    for clase in np.unique(y_original):
        indices_clase = np.flatnonzero(y_original == clase)
        cantidad = min(MAX_PER_CLASS, len(indices_clase))
        seleccion.extend(
            rng.choice(indices_clase, cantidad, replace=False).tolist()
        )

    indices = np.sort(np.asarray(seleccion, dtype=np.int64))
    y = y_original[indices]
    X = np.asarray(f[X_KEY][indices], dtype=np.float32)

    z = np.asarray(f[Z_KEY][indices]) if Z_KEY else None

print("\nMuestras seleccionadas:", len(y))
print("Forma inicial X:", X.shape)
print("Clases originales:", np.unique(y))

# -------------------- VALIDAR Y PREPROCESAR ------------------

if X.shape[-1] == 2:
    X = np.transpose(X, (0, 2, 1))
elif X.shape[1] == 2:
    pass
else:
    raise ValueError(
        "No se reconoce el formato I/Q. "
        f"Se esperaba (N,L,2) o (N,2,L), pero X tiene forma {X.shape}."
    )

if not np.isfinite(X).all():
    raise ValueError("X contiene NaN o infinito; no se entrenará.")

paso = max(1, int(np.ceil(X.shape[-1] / SAMPLE_LENGTH)))
X = X[:, :, ::paso][:, :, :SAMPLE_LENGTH]

if X.shape[-1] < 16:
    raise ValueError(f"Señal demasiado corta tras el muestreo: {X.shape}")

media = X.mean(axis=(1, 2), keepdims=True)
desv = X.std(axis=(1, 2), keepdims=True)
X = (X - media) / np.maximum(desv, 1e-6)
X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

clases_originales = np.unique(y)
mapa_clases = {int(c): i for i, c in enumerate(clases_originales)}
y = np.asarray([mapa_clases[int(c)] for c in y], dtype=np.int64)
N_CLASES = len(clases_originales)

snr = None
if z is not None:
    if z.ndim == 1:
        z_aux = z
    elif z.ndim == 2 and z.shape[1] == 1:
        z_aux = z[:, 0]
    else:
        z_aux = None

    if z_aux is not None:
        try:
            z_aux = z_aux.astype(float)
            if np.isfinite(z_aux).all():
                snr = z_aux
        except (ValueError, TypeError):
            snr = None

print("Forma final X:", X.shape)
print("Número de clases:", N_CLASES)
print("SNR utilizable:", snr is not None)
print("Validación de datos: correcta")

# -------------------- MODELO 1: CNN BASE ---------------------

class CNNBase(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv1d(2, 32, 7, padding=3),
            nn.BatchNorm1d(32), nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(32, 64, 5, padding=2),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, 3, padding=1),
            nn.BatchNorm1d(128), nn.ReLU(),
            nn.AdaptiveAvgPool1d(1)
        )
        self.fc = nn.Linear(128, n_clases)

    def forward(self, x):
        return self.fc(self.features(x).squeeze(-1))

# -------------------- BLOQUE SQUEEZE-EXCITATION --------------

class SEBlock(nn.Module):
    def __init__(self, canales, reduccion=8):
        super().__init__()
        ocultos = max(4, canales // reduccion)
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(canales, ocultos),
            nn.ReLU(),
            nn.Linear(ocultos, canales),
            nn.Sigmoid()
        )

    def forward(self, x):
        pesos = self.fc(self.pool(x).squeeze(-1)).unsqueeze(-1)
        return x * pesos

# -------------------- MODELO 2: CNN DILATADA + SE ------------

class CNNDilatadaSE(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.conv1 = nn.Sequential(
            nn.Conv1d(2, 32, 5, padding=2, dilation=1),
            nn.BatchNorm1d(32), nn.ReLU()
        )
        self.se1 = SEBlock(32)

        self.conv2 = nn.Sequential(
            nn.Conv1d(32, 64, 3, padding=2, dilation=2),
            nn.BatchNorm1d(64), nn.ReLU()
        )
        self.se2 = SEBlock(64)

        self.conv3 = nn.Sequential(
            nn.Conv1d(64, 128, 3, padding=4, dilation=4),
            nn.BatchNorm1d(128), nn.ReLU()
        )
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Linear(128, n_clases)

    def forward(self, x):
        x = self.se1(self.conv1(x))
        x = self.se2(self.conv2(x))
        x = self.conv3(x)
        return self.fc(self.pool(x).squeeze(-1))

# -------------------- MODELO 3: CNN-TRANSFORMER --------------

class CNNTransformer(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv1d(2, 64, 5, padding=2),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.MaxPool1d(2),
            nn.Conv1d(64, 64, 3, padding=1),
            nn.BatchNorm1d(64), nn.ReLU()
        )
        capa = nn.TransformerEncoderLayer(
            d_model=64,
            nhead=4,
            dim_feedforward=128,
            dropout=0.1,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(capa, num_layers=2)
        self.fc = nn.Sequential(
            nn.LayerNorm(64),
            nn.Linear(64, n_clases)
        )

    def forward(self, x):
        x = self.conv(x).transpose(1, 2)
        x = self.transformer(x)
        x = x.mean(dim=1)
        return self.fc(x)

modelos = {
    "CNN base": CNNBase(N_CLASES),
    "CNN dilatada + SE": CNNDilatadaSE(N_CLASES),
    "CNN-Transformer": CNNTransformer(N_CLASES)
}

print("\nModelos creados:")
for nombre, modelo in modelos.items():
    n = sum(p.numel() for p in modelo.parameters() if p.requires_grad)
    print(f"  {nombre}: {n:,} parámetros entrenables")

# -------------------- PARTICIONES COMUNES --------------------

idx = np.arange(len(y))

idx_trainval, idx_test = train_test_split(
    idx, test_size=TEST_SIZE, random_state=SEED, stratify=y
)
val_relativo = VAL_SIZE / (1.0 - TEST_SIZE)

idx_train, idx_val = train_test_split(
    idx_trainval,
    test_size=val_relativo,
    random_state=SEED,
    stratify=y[idx_trainval]
)

def crear_loader(indices, shuffle=False):
    ds = TensorDataset(
        torch.from_numpy(X[indices].astype(np.float32)),
        torch.from_numpy(y[indices].astype(np.int64))
    )
    return DataLoader(
        ds, batch_size=BATCH_SIZE, shuffle=shuffle,
        num_workers=0, pin_memory=(DEVICE.type == "cuda")
    )

train_loader = crear_loader(idx_train, True)
val_loader = crear_loader(idx_val)
test_loader = crear_loader(idx_test)

print("\nParticiones:")
print("Train:", len(idx_train), "| Validación:", len(idx_val),
      "| Test:", len(idx_test))

# -------------------- FUNCIONES DE ENTRENAMIENTO --------------

def ejecutar_epoca(modelo, loader, criterio, optimizador=None):
    entrenando = optimizador is not None
    modelo.train(entrenando)
    loss_total = 0.0
    reales, predichos = [], []

    contexto = torch.enable_grad() if entrenando else torch.no_grad()
    with contexto:
        for xb, yb in loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)

            if entrenando:
                optimizador.zero_grad(set_to_none=True)

            logits = modelo(xb)
            loss = criterio(logits, yb)

            if entrenando:
                loss.backward()
                optimizador.step()

            loss_total += loss.item() * len(yb)
            predichos.extend(logits.argmax(1).detach().cpu().numpy())
            reales.extend(yb.detach().cpu().numpy())

    return (
        loss_total / len(loader.dataset),
        np.asarray(reales),
        np.asarray(predichos)
    )

def entrenar_evaluar(nombre, modelo):
    modelo = modelo.to(DEVICE)
    criterio = nn.CrossEntropyLoss()
    optimizador = torch.optim.AdamW(
        modelo.parameters(), lr=1e-3, weight_decay=1e-4
    )
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizador, mode="min", patience=1, factor=0.5
    )

    mejor_loss = float("inf")
    mejor_estado = None
    contador = 0
    historial = []

    print("\n" + "=" * 60)
    print("ENTRENANDO:", nombre)
    print("=" * 60)

    for epoca in range(1, EPOCHS + 1):
        t0 = time.perf_counter()

        loss_train, yr, yp = ejecutar_epoca(
            modelo, train_loader, criterio, optimizador
        )
        loss_val, yr_val, yp_val = ejecutar_epoca(
            modelo, val_loader, criterio
        )
        scheduler.step(loss_val)

        acc_train = accuracy_score(yr, yp)
        acc_val = accuracy_score(yr_val, yp_val)

        historial.append({
            "epoca": epoca,
            "loss_train": loss_train,
            "loss_val": loss_val,
            "accuracy_train": acc_train,
            "accuracy_val": acc_val
        })

        print(
            f"Época {epoca}/{EPOCHS} | "
            f"loss train={loss_train:.4f} | acc train={acc_train:.4f} | "
            f"loss val={loss_val:.4f} | acc val={acc_val:.4f} | "
            f"tiempo={time.perf_counter()-t0:.1f}s"
        )

        if loss_val < mejor_loss:
            mejor_loss = loss_val
            mejor_estado = {
                k: v.detach().cpu().clone()
                for k, v in modelo.state_dict().items()
            }
            contador = 0
        else:
            contador += 1
            if contador >= PATIENCE:
                print("Parada temprana.")
                break

    if mejor_estado is None:
        raise RuntimeError("No se ha guardado un estado válido.")

    modelo.load_state_dict(mejor_estado)
    modelo.eval()

    ejemplo = torch.from_numpy(
        X[idx_test[:1]].astype(np.float32)
    ).to(DEVICE)

    with torch.no_grad():
        for _ in range(5):
            modelo(ejemplo)
        if DEVICE.type == "cuda":
            torch.cuda.synchronize()

        repeticiones = 50
        t0 = time.perf_counter()
        for _ in range(repeticiones):
            modelo(ejemplo)
        if DEVICE.type == "cuda":
            torch.cuda.synchronize()

    latencia_ms = (time.perf_counter() - t0) * 1000 / repeticiones

    loss_test, y_real, y_pred = ejecutar_epoca(
        modelo, test_loader, criterio
    )

    precision, recall, f1, _ = precision_recall_fscore_support(
        y_real, y_pred, average="weighted", zero_division=0
    )

    resultado = {
        "modelo": nombre,
        "accuracy": accuracy_score(y_real, y_pred),
        "precision_weighted": precision,
        "recall_weighted": recall,
        "f1_weighted": f1,
        "loss_test": loss_test,
        "latencia_ms_por_ejemplo": latencia_ms
    }

    return modelo, resultado, y_real, y_pred, pd.DataFrame(historial)

# -------------------- EJECUTAR LOS TRES MODELOS --------------

resultados = []
predicciones = {}
historiales = {}
modelos_finales = {}
errores = {}

for nombre, modelo in modelos.items():
    try:
        modelo, res, y_real, y_pred, hist = entrenar_evaluar(
            nombre, modelo
        )
        resultados.append(res)
        predicciones[nombre] = (y_real, y_pred)
        historiales[nombre] = hist
        modelos_finales[nombre] = modelo
    except Exception as e:
        errores[nombre] = f"{type(e).__name__}: {e}"
        print(f"\nERROR en {nombre}: {errores[nombre]}")
        if DEVICE.type == "cuda":
            torch.cuda.empty_cache()

tabla = pd.DataFrame(resultados)

print("\n" + "=" * 65)
print("RESULTADOS REALES EN EL CONJUNTO DE PRUEBA")
print("=" * 65)

if tabla.empty:
    print("No se completó ningún modelo. Revisa los errores anteriores.")
else:
    print(tabla.sort_values("accuracy", ascending=False).to_string(index=False))

# -------------------- MATRICES DE CONFUSIÓN ------------------

if predicciones:
    nombres_clases = [str(c) for c in clases_originales]
    etiquetas = np.arange(N_CLASES)

    for nombre, (yr, yp) in predicciones.items():
        cm = confusion_matrix(yr, yp, labels=etiquetas)
        fig, ax = plt.subplots(figsize=(9, 7))
        disp = ConfusionMatrixDisplay(
            confusion_matrix=cm,
            display_labels=nombres_clases
        )
        disp.plot(ax=ax, xticks_rotation="vertical",
                  colorbar=False, values_format="d")
        ax.set_title(f"PROJENER — Matriz de confusión: {nombre}")
        plt.tight_layout()
        plt.show()

        print("\nInforme por clase:", nombre)
        print(classification_report(
            yr, yp, labels=etiquetas,
            target_names=nombres_clases, zero_division=0
        ))

# -------------------- COMPARACIÓN GRÁFICA --------------------

if not tabla.empty:
    tabla.set_index("modelo")["accuracy"].plot(
        kind="bar", figsize=(9, 4), ylim=(0, 1),
        title="PROJENER — Accuracy en test"
    )
    plt.ylabel("Accuracy")
    plt.tight_layout()
    plt.show()

    tabla.set_index("modelo")["latencia_ms_por_ejemplo"].plot(
        kind="bar", figsize=(9, 4),
        title="PROJENER — Latencia de inferencia"
    )
    plt.ylabel("Milisegundos por ejemplo")
    plt.tight_layout()
    plt.show()

# -------------------- MÉTRICAS POR SNR -----------------------

tabla_snr = pd.DataFrame()

if snr is not None and len(snr) == len(y):
    filas = []
    snr_test = snr[idx_test]

    for nombre, (yr, yp) in predicciones.items():
        for nivel in np.unique(snr_test):
            mascara = snr_test == nivel
            if mascara.sum() == 0:
                continue

            filas.append({
                "modelo": nombre,
                "SNR_dB": float(nivel),
                "n_ejemplos": int(mascara.sum()),
                "accuracy": accuracy_score(yr[mascara], yp[mascara])
            })

    tabla_snr = pd.DataFrame(filas)

    if not tabla_snr.empty:
        print("\nMÉTRICAS POR SNR")
        print(tabla_snr.to_string(index=False))

        for nombre, grupo in tabla_snr.groupby("modelo"):
            plt.plot(
                grupo["SNR_dB"], grupo["accuracy"],
                marker="o", label=nombre
            )

        plt.xlabel("SNR (dB)")
        plt.ylabel("Accuracy")
        plt.title("PROJENER — Accuracy por SNR")
        plt.ylim(0, 1)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()
else:
    print(
        "\nNo se genera la evaluación por SNR: Z no contiene "
        "un vector numérico compatible."
    )

# -------------------- EXPORTAR RESULTADOS --------------------

OUT = Path("/content/resultados_PROJENER")
OUT.mkdir(parents=True, exist_ok=True)

if not tabla.empty:
    tabla.to_csv(OUT / "comparacion_modelos.csv", index=False)

if not tabla_snr.empty:
    tabla_snr.to_csv(OUT / "metricas_por_SNR.csv", index=False)

for nombre, hist in historiales.items():
    seguro = (
        nombre.lower().replace(" ", "_")
        .replace("+", "plus").replace("-", "_")
    )
    hist.to_csv(OUT / f"historial_{seguro}.csv", index=False)

for nombre, modelo in modelos_finales.items():
    seguro = (
        nombre.lower().replace(" ", "_")
        .replace("+", "plus").replace("-", "_")
    )
    torch.save(modelo.cpu().state_dict(), OUT / f"{seguro}.pth")
    modelo.to(DEVICE)

config = {
    "dataset": os.path.basename(DATA_PATH),
    "seed": SEED,
    "max_per_class": MAX_PER_CLASS,
    "sample_length": int(X.shape[-1]),
    "clases_originales": [str(c) for c in clases_originales],
    "device": str(DEVICE),
    "epochs_max": EPOCHS,
    "batch_size": BATCH_SIZE,
    "modelos_completados": list(modelos_finales.keys()),
    "errores": errores
}

with open(OUT / "configuracion.json", "w", encoding="utf-8") as f:
    json.dump(config, f, ensure_ascii=False, indent=2)

ZIP_PATH = "/content/resultados_PROJENER.zip"
with zipfile.ZipFile(ZIP_PATH, "w", zipfile.ZIP_DEFLATED) as zf:
    for archivo in OUT.iterdir():
        if archivo.is_file():
            zf.write(archivo, arcname=archivo.name)

print("\n" + "=" * 65)
print("PROCESO FINALIZADO")
print("Modelos completados:", list(modelos_finales.keys()))
print("Resultados:", OUT)
print("Paquete ZIP:", ZIP_PATH)
if errores:
    print("Modelos con errores:", errores)

try:
    from google.colab import files
    files.download(ZIP_PATH)
except Exception:
    print("Descarga manual disponible en:", ZIP_PATH)
```

```python


import os, time, random, json, zipfile, tarfile, pickle
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support,
    classification_report, confusion_matrix, ConfusionMatrixDisplay
)

# -------------------- CONFIGURACIÓN --------------------------

SEED = 42
MAX_PER_CLASS = 500
SAMPLE_LENGTH = 128  # RML2016.10a tiene longitud fija de 128 muestras por defecto
EPOCHS = 10
BATCH_SIZE = 128
PATIENCE = 3
TEST_SIZE = 0.20
VAL_SIZE = 0.15

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 65)
print("PROJENER — EXPERIMENTO COMPARATIVO (Dataset RML2016.10a)")
print("=" * 65)
print("Dispositivo:", DEVICE)
if DEVICE.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
print("PyTorch:", torch.__version__)

# -------------------- LOCALIZAR Y DESCOMPRIMIR DATASET -------

ruta_tar = Path("/content/RML2016.10a.tar.bz2")
if not ruta_tar.exists():
    archivos_tar = list(Path("/content").glob("*.tar.bz2"))
    if archivos_tar:
        ruta_tar = archivos_tar[0]
    else:
        try:
            from google.colab import files
            print("\nSube el archivo RML2016.10a.tar.bz2:")
            subidos = files.upload()
            ruta_tar = Path("/content") / list(subidos.keys())[0]
        except Exception as e:
            raise FileNotFoundError("No se encuentra el archivo RML2016.10a.tar.bz2.") from e

print("\nDescomprimiendo dataset...")
with tarfile.open(ruta_tar, "r:bz2") as tar:
    tar.extractall(path="/content")
print("Descompresión completada.")

# Buscar el archivo .pkl resultante
archivos_pkl = list(Path("/content").glob("*.pkl"))
if not archivos_pkl:
    # Buscar recursivamente si se extrajo dentro de una carpeta
    archivos_pkl = list(Path("/content").rglob("*.pkl"))

if not archivos_pkl:
    raise FileNotFoundError("No se encontró el archivo .pkl tras descomprimir el tar.bz2.")

DATA_PATH = str(archivos_pkl[0])
print("Archivo Pickle detectado:", DATA_PATH)

# -------------------- CARGA Y EXTRACCIÓN RML2016.10a ---------

print("\nCargando diccionario de datos (Pickle)...")
with open(DATA_PATH, "rb") as f:
    # RML2016.10a usa encoding 'latin1' para compatibilidad con Python 3
    dataset_dict = pickle.load(f, encoding='latin1')

# El formato típico de RML2016.10a es un diccionario {(modulacion, snr): array(N, 2, 128)}
X_lista = []
y_lista = []
snr_lista = []

# Obtener lista de modulaciones únicas a partir de las claves del diccionario
modulaciones = sorted(list(set([key[0] for key in dataset_dict.keys()])))
mapa_modulaciones = {mod: i for i, mod in enumerate(modulaciones)}

print(f"Modulaciones detectadas ({len(modulaciones)}):", modulaciones)

for (mod, snr_val), datos in dataset_dict.items():
    clase_idx = mapa_modulaciones[mod]

    # Aplicar límite de muestras por clase/snr para equilibrar y agilizar
    n_muestras = datos.shape[0]
    if MAX_PER_CLASS and n_muestras > MAX_PER_CLASS:
        indices_sel = np.random.choice(n_muestras, MAX_PER_CLASS, replace=False)
        datos = datos[indices_sel]
        n_muestras = MAX_PER_CLASS

    X_lista.append(datos)
    y_lista.extend([clase_idx] * n_muestras)
    snr_lista.extend([snr_val] * n_muestras)

X = np.concatenate(X_lista, axis=0).astype(np.float32)
y = np.array(y_lista, dtype=np.int64)
snr = np.array(snr_lista, dtype=np.float32)

N_CLASES = len(modulaciones)
print("Forma inicial X:", X.shape)
print("Número total de muestras cargadas:", len(y))

# -------------------- VALIDAR Y PREPROCESAR ------------------

# RML2016.10a viene típicamente en formato (N, 2, 128) -> (Canales I/Q, Longitud)
if X.shape[1] != 2 and X.shape[-1] == 2:
    X = np.transpose(X, (0, 2, 1))

if not np.isfinite(X).all():
    raise ValueError("X contiene NaN o infinitos.")

# Normalización independiente por muestra
media = X.mean(axis=(1, 2), keepdims=True)
desv = X.std(axis=(1, 2), keepdims=True)
X = (X - media) / np.maximum(desv, 1e-6)
X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

print("Forma final X preprocesada:", X.shape)
print("Validación de datos: correcta")

# -------------------- MODELO 1: CNN BASE ---------------------

class CNNBase(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv1d(2, 32, 7, padding=3),
            nn.BatchNorm1d(32), nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(32, 64, 5, padding=2),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, 3, padding=1),
            nn.BatchNorm1d(128), nn.ReLU(),
            nn.AdaptiveAvgPool1d(1)
        )
        self.fc = nn.Linear(128, n_clases)

    def forward(self, x):
        return self.fc(self.features(x).squeeze(-1))

# -------------------- BLOQUE SQUEEZE-EXCITATION --------------

class SEBlock(nn.Module):
    def __init__(self, canales, reduccion=8):
        super().__init__()
        ocultos = max(4, canales // reduccion)
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(canales, ocultos),
            nn.ReLU(),
            nn.Linear(ocultos, canales),
            nn.Sigmoid()
        )

    def forward(self, x):
        pesos = self.fc(self.pool(x).squeeze(-1)).unsqueeze(-1)
        return x * pesos

# -------------------- MODELO 2: CNN DILATADA + SE ------------

class CNNDilatadaSE(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.conv1 = nn.Sequential(
            nn.Conv1d(2, 32, 5, padding=2, dilation=1),
            nn.BatchNorm1d(32), nn.ReLU()
        )
        self.se1 = SEBlock(32)

        self.conv2 = nn.Sequential(
            nn.Conv1d(32, 64, 3, padding=2, dilation=2),
            nn.BatchNorm1d(64), nn.ReLU()
        )
        self.se2 = SEBlock(64)

        self.conv3 = nn.Sequential(
            nn.Conv1d(64, 128, 3, padding=4, dilation=4),
            nn.BatchNorm1d(128), nn.ReLU()
        )
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Linear(128, n_clases)

    def forward(self, x):
        x = self.se1(self.conv1(x))
        x = self.se2(self.conv2(x))
        x = self.conv3(x)
        return self.fc(self.pool(x).squeeze(-1))

# -------------------- MODELO 3: CNN-TRANSFORMER --------------

class CNNTransformer(nn.Module):
    def __init__(self, n_clases):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv1d(2, 64, 5, padding=2),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.MaxPool1d(2),
            nn.Conv1d(64, 64, 3, padding=1),
            nn.BatchNorm1d(64), nn.ReLU()
        )
        capa = nn.TransformerEncoderLayer(
            d_model=64,
            nhead=4,
            dim_feedforward=128,
            dropout=0.1,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(capa, num_layers=2)
        self.fc = nn.Sequential(
            nn.LayerNorm(64),
            nn.Linear(64, n_clases)
        )

    def forward(self, x):
        x = self.conv(x).transpose(1, 2)
        x = self.transformer(x)
        x = x.mean(dim=1)
        return self.fc(x)

modelos = {
    "CNN base": CNNBase(N_CLASES),
    "CNN dilatada + SE": CNNDilatadaSE(N_CLASES),
    "CNN-Transformer": CNNTransformer(N_CLASES)
}

print("\nModelos creados:")
for nombre, modelo in modelos.items():
    n = sum(p.numel() for p in modelo.parameters() if p.requires_grad)
    print(f"  {nombre}: {n:,} parámetros entrenables")

# -------------------- PARTICIONES COMUNES --------------------

idx = np.arange(len(y))

idx_trainval, idx_test = train_test_split(
    idx, test_size=TEST_SIZE, random_state=SEED, stratify=y
)
val_relativo = VAL_SIZE / (1.0 - TEST_SIZE)

idx_train, idx_val = train_test_split(
    idx_trainval,
    test_size=val_relativo,
    random_state=SEED,
    stratify=y[idx_trainval]
)

def crear_loader(indices, shuffle=False):
    ds = TensorDataset(
        torch.from_numpy(X[indices].astype(np.float32)),
        torch.from_numpy(y[indices].astype(np.int64))
    )
    return DataLoader(
        ds, batch_size=BATCH_SIZE, shuffle=shuffle,
        num_workers=0, pin_memory=(DEVICE.type == "cuda")
    )

train_loader = crear_loader(idx_train, True)
val_loader = crear_loader(idx_val)
test_loader = crear_loader(idx_test)

print("\nParticiones:")
print("Train:", len(idx_train), "| Validación:", len(idx_val), "| Test:", len(idx_test))

# -------------------- FUNCIONES DE ENTRENAMIENTO --------------

def ejecutar_epoca(modelo, loader, criterio, optimizador=None):
    entrenando = optimizador is not None
    modelo.train(entrenando)
    loss_total = 0.0
    reales, predichos = [], []

    contexto = torch.enable_grad() if entrenando else torch.no_grad()
    with contexto:
        for xb, yb in loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)

            if entrenando:
                optimizador.zero_grad(set_to_none=True)

            logits = modelo(xb)
            loss = criterio(logits, yb)

            if entrenando:
                loss.backward()
                optimizador.step()

            loss_total += loss.item() * len(yb)
            predichos.extend(logits.argmax(1).detach().cpu().numpy())
            reales.extend(yb.detach().cpu().numpy())

    return (
        loss_total / len(loader.dataset),
        np.asarray(reales),
        np.asarray(predichos)
    )

def entrenar_evaluar(nombre, modelo):
    modelo = modelo.to(DEVICE)
    criterio = nn.CrossEntropyLoss()
    optimizador = torch.optim.AdamW(
        modelo.parameters(), lr=1e-3, weight_decay=1e-4
    )
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizador, mode="min", patience=1, factor=0.5
    )

    mejor_loss = float("inf")
    mejor_estado = None
    contador = 0
    historial = []

    print("\n" + "=" * 60)
    print("ENTRENANDO:", nombre)
    print("=" * 60)

    for epoca in range(1, EPOCHS + 1):
        t0 = time.perf_counter()

        loss_train, yr, yp = ejecutar_epoca(
            modelo, train_loader, criterio, optimizador
        )
        loss_val, yr_val, yp_val = ejecutar_epoca(
            modelo, val_loader, criterio
        )
        scheduler.step(loss_val)

        acc_train = accuracy_score(yr, yp)
        acc_val = accuracy_score(yr_val, yp_val)

        historial.append({
            "epoca": epoca,
            "loss_train": loss_train,
            "loss_val": loss_val,
            "accuracy_train": acc_train,
            "accuracy_val": acc_val
        })

        print(
            f"Época {epoca}/{EPOCHS} | "
            f"loss train={loss_train:.4f} | acc train={acc_train:.4f} | "
            f"loss val={loss_val:.4f} | acc val={acc_val:.4f} | "
            f"tiempo={time.perf_counter()-t0:.1f}s"
        )

        if loss_val < mejor_loss:
            mejor_loss = loss_val
            mejor_estado = {
                k: v.detach().cpu().clone()
                for k, v in modelo.state_dict().items()
            }
            contador = 0
        else:
            contador += 1
            if contador >= PATIENCE:
                print("Parada temprana.")
                break

    if mejor_estado is None:
        raise RuntimeError("No se ha guardado un estado válido.")

    modelo.load_state_dict(mejor_estado)
    modelo.eval()

    ejemplo = torch.from_numpy(
        X[idx_test[:1]].astype(np.float32)
    ).to(DEVICE)

    with torch.no_grad():
        for _ in range(5):
            modelo(ejemplo)
        if DEVICE.type == "cuda":
            torch.cuda.synchronize()

        repeticiones = 50
        t0 = time.perf_counter()
        for _ in range(repeticiones):
            modelo(ejemplo)
        if DEVICE.type == "cuda":
            torch.cuda.synchronize()

    latencia_ms = (time.perf_counter() - t0) * 1000 / repeticiones

    loss_test, y_real, y_pred = ejecutar_epoca(
        modelo, test_loader, criterio
    )

    precision, recall, f1, _ = precision_recall_fscore_support(
        y_real, y_pred, average="weighted", zero_division=0
    )

    resultado = {
        "modelo": nombre,
        "accuracy": accuracy_score(y_real, y_pred),
        "precision_weighted": precision,
        "recall_weighted": recall,
        "f1_weighted": f1,
        "loss_test": loss_test,
        "latencia_ms_por_ejemplo": latencia_ms
    }

    return modelo, resultado, y_real, y_pred, pd.DataFrame(historial)

# -------------------- EJECUTAR LOS TRES MODELOS --------------

resultados = []
predicciones = {}
historiales = {}
modelos_finales = {}
errores = {}

for nombre, modelo in modelos.items():
    try:
        modelo, res, y_real, y_pred, hist = entrenar_evaluar(nombre, modelo)
        resultados.append(res)
        predicciones[nombre] = (y_real, y_pred)
        historiales[nombre] = hist
        modelos_finales[nombre] = modelo
    except Exception as e:
        errores[nombre] = f"{type(e).__name__}: {e}"
        print(f"\nERROR en {nombre}: {errores[nombre]}")
        if DEVICE.type == "cuda":
            torch.cuda.empty_cache()

tabla = pd.DataFrame(resultados)

print("\n" + "=" * 65)
print("RESULTADOS REALES EN EL CONJUNTO DE PRUEBA")
print("=" * 65)

if tabla.empty:
    print("No se completó ningún modelo.")
else:
    print(tabla.sort_values("accuracy", ascending=False).to_string(index=False))

# -------------------- MATRICES DE CONFUSIÓN ------------------

if predicciones:
    nombres_clases = modulaciones
    etiquetas = np.arange(N_CLASES)

    for nombre, (yr, yp) in predicciones.items():
        cm = confusion_matrix(yr, yp, labels=etiquetas)
        fig, ax = plt.subplots(figsize=(9, 7))
        disp = ConfusionMatrixDisplay(
            confusion_matrix=cm,
            display_labels=nombres_clases
        )
        disp.plot(ax=ax, xticks_rotation="vertical", colorbar=False, values_format="d")
        ax.set_title(f"PROJENER — Matriz de confusión: {nombre}")
        plt.tight_layout()
        plt.show()

# -------------------- EXPORTAR RESULTADOS --------------------

OUT = Path("/content/resultados_PROJENER")
OUT.mkdir(parents=True, exist_ok=True)

if not tabla.empty:
    tabla.to_csv(OUT / "comparacion_modelos.csv", index=False)

ZIP_PATH = "/content/resultados_PROJENER.zip"
with zipfile.ZipFile(ZIP_PATH, "w", zipfile.ZIP_DEFLATED) as zf:
    for archivo in OUT.iterdir():
        if archivo.is_file():
            zf.write(archivo, arcname=archivo.name)

print("\n" + "=" * 65)
print("PROCESO FINALIZADO CON ÉXITO")
print("Paquete ZIP generado:", ZIP_PATH)

try:
    from google.colab import files
    files.download(ZIP_PATH)
except Exception:
    pass
```
